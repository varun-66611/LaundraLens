```python
"""
Mule-account detection backend (single file, beginner friendly).

What it does when you press F5 in IDLE:
  1. Loads PaySim CSV if it sits next to this file, otherwise generates fake data
  2. Builds a transaction graph (NetworkX)
  3. Finds clusters (Louvain)
  4. Computes features per account
  5. Trains a Random Forest and prints precision / recall
  6. Saves a picture of the network (network.png)
  7. Starts an API at http://127.0.0.1:8000  (docs page: /docs)

Install once (in Command Prompt / Terminal, not in IDLE):
  In Colab: run  !pip install -q fastapi httpx  in the cell above
"""

import os
import sys
import random

import numpy as np
import pandas as pd
import networkx as nx
import joblib
import matplotlib
matplotlib.use("Agg")  # draw to a file, never opens a window (safe for IDLE)
import matplotlib.pyplot as plt

from networkx.algorithms.community import louvain_communities
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

# ----------------------------------------------------------------------
# Settings
# ----------------------------------------------------------------------
BASE_DIR = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()
PAYSIM_FILE = os.path.join(BASE_DIR, "PS_20174392719_1491204439457_log.csv")
MODEL_FILE = os.path.join(BASE_DIR, "model.pkl")
FEATURE_FILE = os.path.join(BASE_DIR, "account_features.csv")
PICTURE_FILE = os.path.join(BASE_DIR, "network.png")

FEATURES = [
    "total_in", "total_out", "in_degree", "out_degree",
    "pass_through", "community_size",
]
SEED = 42


# ----------------------------------------------------------------------
# Step 1: Data
# ----------------------------------------------------------------------
def generate_synthetic(n_normal=1000, n_tx=8000, n_chains=30, n_businesses=15, seed=SEED):
    """Fake data: normal people, busy businesses (decoys), and mule chains."""
    rng = random.Random(seed)
    rows = []  # (nameOrig, nameDest, amount, isFraud)
    mules = set()

    normal = [f"C{i}" for i in range(n_normal)]

    # Normal people paying each other (small amounts, a few big ones)
    for _ in range(n_tx):
        a, b = rng.sample(normal, 2)
        rows.append((a, b, round(rng.lognormvariate(5.5, 1.2), 2), 0))

    # Legit businesses: many customers pay in, many employees get paid out.
    # They look like pass-through accounts, so the model must not flag them.
    for k in range(n_businesses):
        biz = f"B{k}"
        total = 0.0
        for cust in rng.sample(normal, 30):
            amt = round(rng.uniform(100, 2000), 2)
            total += amt
            rows.append((cust, biz, amt, 0))
        employees = rng.sample(normal, 10)
        for emp in employees:
            rows.append((biz, emp, round(total * 0.95 / 10, 2), 0))

    # Mule chains: victim(s) -> m0 -> m1 -> ... -> mN -> someone
    for c in range(n_chains):
        length = rng.randint(5, 10)
        ids = [f"M{c}_{i}" for i in range(length)]
        mules.update(ids)
        amt = rng.uniform(3000, 40000)
        for victim in rng.sample(normal, rng.randint(1, 3)):
            rows.append((victim, ids[0], round(amt / 2, 2), 1))
        for i in range(length - 1):
            rows.append((ids[i], ids[i + 1], round(amt, 2), 1))
            amt *= 0.98  # each mule keeps a small cut
        rows.append((ids[-1], rng.choice(normal), round(amt, 2), 1))

    df = pd.DataFrame(rows, columns=["nameOrig", "nameDest", "amount", "isFraud"])
    return df, mules


def load_paysim(path):
    """Real-ish data. Keeps all fraud rows + a sample of normal rows."""
    df = pd.read_csv(path)
    df = df[df["type"].isin(["TRANSFER", "CASH_OUT"])]
    fraud = df[df["isFraud"] == 1]
    normal = df[df["isFraud"] == 0].sample(50000, random_state=SEED)
    df = pd.concat([fraud, normal])[["nameOrig", "nameDest", "amount", "isFraud"]]
    mules = set(df.loc[df["isFraud"] == 1, "nameOrig"])
    return df.reset_index(drop=True), mules


def get_data():
    if os.path.exists(PAYSIM_FILE):
        print("Using PaySim file:", PAYSIM_FILE)
        return load_paysim(PAYSIM_FILE)
    print("PaySim file not found next to backend.py -> generating synthetic data.")
    return generate_synthetic()


# ----------------------------------------------------------------------
# Step 2 and 3: Graph + clusters
# ----------------------------------------------------------------------
def build_graph(df):
    G = nx.DiGraph()
    grouped = df.groupby(["nameOrig", "nameDest"])["amount"].sum()
    for (sender, receiver), amount in grouped.items():
        if sender != receiver:
            G.add_edge(sender, receiver, amount=float(amount))
    return G


def find_communities(G):
    communities = louvain_communities(G.to_undirected(), seed=SEED)
    community_of = {}
    for i, group in enumerate(communities):
        for account in group:
            community_of[account] = i
    sizes = {i: len(g) for i, g in enumerate(communities)}
    return community_of, sizes


# ----------------------------------------------------------------------
# Step 4: Features
# ----------------------------------------------------------------------
def make_pass_through(total_in, total_out):
    """1.0 = everything that came in went straight out. 0 = one-sided."""
    total_in = np.asarray(total_in, dtype=float)
    total_out = np.asarray(total_out, dtype=float)
    big = np.maximum(total_in, total_out)
    small = np.minimum(total_in, total_out)
    return np.where(big > 0, small / np.where(big > 0, big, 1.0), 0.0)


def build_features(G, community_of, sizes, mules):
    feat = pd.DataFrame(index=list(G.nodes))
    feat["total_in"] = pd.Series(dict(G.in_degree(weight="amount")))
    feat["total_out"] = pd.Series(dict(G.out_degree(weight="amount")))
    feat["in_degree"] = pd.Series(dict(G.in_degree()))
    feat["out_degree"] = pd.Series(dict(G.out_degree()))
    feat["pass_through"] = make_pass_through(feat["total_in"], feat["total_out"])
    feat["community_size"] = [sizes[community_of[a]] for a in feat.index]
    feat["label"] = [1 if a in mules else 0 for a in feat.index]
    feat.index.name = "account"
    return feat


# ----------------------------------------------------------------------
# Step 5: Train
# ----------------------------------------------------------------------
def train_model(feat):
    X = feat[FEATURES]
    y = feat["label"]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=SEED, stratify=y
    )
    model = RandomForestClassifier(
        n_estimators=200, class_weight="balanced", random_state=SEED
    )
    model.fit(X_train, y_train)

    print("\n=== Results on accounts the model has never seen ===")
    print(classification_report(y_test, model.predict(X_test), digits=3, zero_division=0))
    print("Feature importance (what drives the flags):")
    for name, imp in sorted(zip(FEATURES, model.feature_importances_), key=lambda t: -t[1]):
        print(f"  {name:15s} {imp:.3f}")
    return model


# ----------------------------------------------------------------------
# Step 6: Picture
# ----------------------------------------------------------------------
def draw_network(G, flagged_accounts, path):
    # Show the neighbourhood around flagged accounts so the picture stays readable
    keep = set()
    for n in list(flagged_accounts)[:60]:
        if n in G:
            keep.add(n)
            keep.update(G.predecessors(n))
            keep.update(G.successors(n))
    sub = G.subgraph(list(keep)[:300])
    if sub.number_of_nodes() == 0:
        print("Nothing to draw.")
        return
    colors = ["red" if n in flagged_accounts else "lightblue" for n in sub.nodes]
    plt.figure(figsize=(10, 10))
    nx.draw(sub, pos=nx.spring_layout(sub, seed=SEED), node_color=colors,
            node_size=50, arrows=True, width=0.5)
    plt.title("Red = flagged mule accounts")
    plt.savefig(path, dpi=120, bbox_inches="tight")
    plt.close()
    print("Saved network picture:", path)


# ----------------------------------------------------------------------
# Pipeline: run everything once
# ----------------------------------------------------------------------
STATE = {"model": None, "features": None}


def run_pipeline():
    df, mules = get_data()
    print("Transactions:", len(df), "| mule accounts:", len(mules))

    G = build_graph(df)
    print("Accounts:", G.number_of_nodes(), "| Transfer links:", G.number_of_edges())

    community_of, sizes = find_communities(G)
    print("Clusters found:", len(sizes))

    feat = build_features(G, community_of, sizes, mules)
    model = train_model(feat)

    feat["risk"] = model.predict_proba(feat[FEATURES])[:, 1]
    flagged = set(feat.index[feat["risk"] > 0.5])
    print("\nAccounts flagged (risk > 0.5):", len(flagged))
    draw_network(G, flagged, PICTURE_FILE)

    joblib.dump(model, MODEL_FILE)
    feat.to_csv(FEATURE_FILE)
    STATE["model"] = model
    STATE["features"] = feat


# ----------------------------------------------------------------------
# Step 7: API
# ----------------------------------------------------------------------
app = FastAPI(title="Mule Account Detector")


class AccountInput(BaseModel):
    total_in: float
    total_out: float
    in_degree: int
    out_degree: int
    community_size: int


def score_row(row_dict):
    row = dict(row_dict)
    row["pass_through"] = float(make_pass_through(row["total_in"], row["total_out"]))
    x = pd.DataFrame([row])[FEATURES]
    return float(STATE["model"].predict_proba(x)[0][1]), row["pass_through"]


@app.get("/")
def home():
    return {"status": "running", "docs": "/docs"}


@app.post("/score")
def score(account: AccountInput):
    risk, pt = score_row(account.model_dump() if hasattr(account, 'model_dump') else account.dict())
    return {"risk": round(risk, 3), "flagged": risk > 0.5, "pass_through": round(pt, 3)}


@app.get("/account/{account_id}")
def account_info(account_id: str):
    feat = STATE["features"]
    if account_id not in feat.index:
        raise HTTPException(status_code=404, detail="Account not found")
    row = feat.loc[account_id]
    return {
        "account": account_id,
        "risk": round(float(row["risk"]), 3),
        "flagged": bool(row["risk"] > 0.5),
        "features": {f: float(row[f]) for f in FEATURES},
    }


@app.get("/flagged")
def flagged(limit: int = 20):
    feat = STATE["features"]
    top = feat.sort_values("risk", ascending=False).head(limit)
    return [{"account": a, "risk": round(float(r), 3)} for a, r in top["risk"].items()]


# ----------------------------------------------------------------------
# Run + demo (Colab-safe: no server needed, the API is tested in-notebook)
# ----------------------------------------------------------------------
run_pipeline()

from fastapi.testclient import TestClient
client = TestClient(app)

print("\n--- API test 1: score a mule-like account ---")
print(client.post("/score", json={"total_in": 50000, "total_out": 49000,
      "in_degree": 1, "out_degree": 1, "community_size": 7}).json())

print("\n--- API test 2: score a normal account ---")
print(client.post("/score", json={"total_in": 300, "total_out": 120,
      "in_degree": 8, "out_degree": 7, "community_size": 300}).json())

print("\n--- API test 3: top 5 risky accounts ---")
top = client.get("/flagged?limit=5").json()
print(top)

print("\n--- API test 4: details of one account ---")
print(client.get("/account/" + top[0]["account"]).json())

try:
    from IPython.display import Image, display
    display(Image(PICTURE_FILE))
except Exception as e:
    print("Picture saved as", PICTURE_FILE, "(open it from the folder icon on the left)")
```

```python
# ======================================================================
# CELL 2: Detailed mule-account report (run AFTER the first code cell)
# Gives: (1) a table of every flagged account with full details,
#        (2) an annotated network image of the riskiest chains,
#        (3) the transaction history of any single account.
# ======================================================================
import re
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from IPython.display import display, Image

# Make sure the model from the first cell exists
if STATE.get("model") is None:
    run_pipeline()

THRESHOLD = 0.5      # risk above this = red flag
TOP_N_TABLE = 30     # how many flagged accounts to show in the table
TOP_N_CHAINS = 4     # how many suspicious chains to draw in the image

# Same data as before (the data generator uses a fixed seed, so it is identical)
df_all, _ = get_data()
G_all = build_graph(df_all)
feat = STATE["features"].copy()


# ---------------- account type (behaviour-based guess) ----------------
def account_type(acc, in_deg):
    if re.fullmatch(r"M\d+", str(acc)):          # PaySim merchant ids look like M123456
        return "Merchant"
    if in_deg >= 25:                              # many different senders paying in
        return "Business / Seller"
    return "Personal"


feat["type"] = [account_type(a, d) for a, d in zip(feat.index, feat["in_degree"])]

# ---------------- previous history numbers per account ----------------
g_sent = df_all.groupby("nameOrig")
g_recv = df_all.groupby("nameDest")
idx = feat.index
feat["tx_count"] = (g_sent.size().reindex(idx, fill_value=0)
                    + g_recv.size().reindex(idx, fill_value=0))
feat["fraud_txns"] = (g_sent["isFraud"].sum().reindex(idx, fill_value=0)
                      + g_recv["isFraud"].sum().reindex(idx, fill_value=0))
feat["largest_tx"] = np.maximum(g_sent["amount"].max().reindex(idx, fill_value=0),
                                g_recv["amount"].max().reindex(idx, fill_value=0))


def risk_level(p):
    if p >= 0.90:
        return "\U0001F6A9 HIGH"
    if p >= 0.70:
        return "\U0001F6A9 MEDIUM"
    return "\U0001F6A9 LOW"


def explain(r):
    why = []
    if r["total_in"] > 0 and r["pass_through"] >= 0.9:
        why.append(f"passes on {r['pass_through'] * 100:.0f}% of incoming money")
    if r["in_degree"] <= 2 and r["out_degree"] <= 2:
        why.append("very few counterparties")
    if r["community_size"] <= 15:
        why.append(f"tight cluster of {int(r['community_size'])} accounts")
    if r["fraud_txns"] > 0:
        why.append(f"{int(r['fraud_txns'])} past fraud-linked transaction(s)")
    if not why:
        why.append("overall pattern similar to known mule accounts")
    return "; ".join(why)


flagged_df = feat[feat["risk"] > THRESHOLD].sort_values("risk", ascending=False)

# ---------------- (1) the table ----------------
rows = []
for acc, r in flagged_df.iterrows():
    rows.append({
        "Account": acc,
        "Flag": risk_level(r["risk"]),
        "Detection %": f"{r['risk'] * 100:.1f}%",
        "Account Type": r["type"],
        "Amount Received": f"{r['total_in']:,.2f}",
        "Amount Transferred": f"{r['total_out']:,.2f}",
        "Sent To (no. of accounts)": int(r["out_degree"]),
        "Received From (no. of accounts)": int(r["in_degree"]),
        "Pass-through %": f"{r['pass_through'] * 100:.0f}%",
        "Cluster Size": int(r["community_size"]),
        "Total Past Txns": int(r["tx_count"]),
        "Past Fraud-linked Txns": int(r["fraud_txns"]),
        "Largest Txn": f"{r['largest_tx']:,.2f}",
        "Why Flagged": explain(r),
    })
report = pd.DataFrame(rows)
report.to_csv("flagged_accounts_report.csv", index=False)


def row_color(row):
    level = row["Flag"].split()[-1]
    color = {"HIGH": "#ffb3b3", "MEDIUM": "#ffd9b3", "LOW": "#fff2b3"}.get(level, "")
    return [f"background-color: {color}"] * len(row)


print(f"RED-FLAGGED ACCOUNTS: {len(report)}   |   "
      f"money moved by flagged accounts: {flagged_df['total_out'].sum():,.0f}")
print(f"Showing top {min(TOP_N_TABLE, len(report))} (full list saved to flagged_accounts_report.csv)\n")
shown = report.head(TOP_N_TABLE)
try:
    display(shown.style.apply(row_color, axis=1).hide(axis="index"))
except Exception:
    display(shown)


# ---------------- (2) annotated network image ----------------
def short(x):
    return f"{x / 1000:.1f}k" if x >= 1000 else f"{x:.0f}"


def chain_layout(sub):
    """Left-to-right layout so money flow reads like a chain."""
    try:
        pos = {}
        for x, gen in enumerate(nx.topological_generations(sub)):
            n = len(gen)
            for j, node in enumerate(sorted(gen)):
                pos[node] = (x * 1.6, (j - (n - 1) / 2) * 1.5)
        return pos
    except Exception:
        return nx.spring_layout(sub, seed=SEED)


def draw_report(path="mule_report.png"):
    flagged_set = set(flagged_df.index) & set(G_all.nodes)
    if not flagged_set:
        print("No flagged accounts to draw.")
        return None
    comps = [c for c in nx.weakly_connected_components(G_all.subgraph(flagged_set))]
    comps.sort(key=lambda c: (flagged_df.loc[list(c), "risk"].mean(),
                              flagged_df.loc[list(c), "total_out"].sum()), reverse=True)
    comps = comps[:TOP_N_CHAINS]

    fig, axes = plt.subplots(len(comps), 1, figsize=(24, 3.6 * len(comps)))
    axes = np.atleast_1d(axes)
    for ax, comp in zip(axes, comps):
        edges = [(u, v) for u, v in G_all.edges()
                 if u in comp or v in comp]
        sub = G_all.edge_subgraph(edges).copy()
        pos = chain_layout(sub)
        node_colors, node_sizes = [], []
        for n in sub.nodes:
            if n in flagged_set:
                node_colors.append(plt.cm.Reds(0.35 + 0.65 * float(feat.loc[n, "risk"])))
                node_sizes.append(1500)
            else:
                node_colors.append("lightblue")
                node_sizes.append(900)
        nx.draw_networkx_edges(sub, pos, ax=ax, arrows=True, arrowsize=18,
                               node_size=node_sizes, width=1.5, edge_color="gray")
        nx.draw_networkx_nodes(sub, pos, ax=ax, node_color=node_colors,
                               node_size=node_sizes, edgecolors="black")
        nx.draw_networkx_edge_labels(
            sub, pos, ax=ax, font_size=9,
            edge_labels={(u, v): short(d["amount"]) for u, v, d in sub.edges(data=True)})
        for n, (x, y) in pos.items():
            r = feat.loc[n]
            if n in flagged_set:
                text = (f"RED FLAG {n}\nrisk {r['risk'] * 100:.0f}% | {r['type']}\n"
                        f"in {short(r['total_in'])} / out {short(r['total_out'])}\n"
                        f"from {int(r['in_degree'])} acc, to {int(r['out_degree'])} acc")
                box = dict(boxstyle="round", fc="#ffe6e6", ec="red")
            else:
                text = f"{n}\n{r['type']}\nrisk {r['risk'] * 100:.0f}%"
                box = dict(boxstyle="round", fc="#eef6ff", ec="steelblue")
            ax.text(x, y - 0.3, text, ha="center", va="top", fontsize=9.5, bbox=box)
        xs = [p[0] for p in pos.values()]
        ys = [p[1] for p in pos.values()]
        ax.set_xlim(min(xs) - 1, max(xs) + 1)
        ax.set_ylim(min(ys) - 1.9, max(ys) + 0.6)
        ax.set_title(f"Suspicious chain {comps.index(comp) + 1}: "
                     f"{len(comp)} flagged accounts, "
                     f"avg risk {flagged_df.loc[list(comp), 'risk'].mean() * 100:.0f}% "
                     f"(red = flagged, blue = other accounts linked to them)")
        ax.axis("off")
    plt.tight_layout()
    plt.savefig(path, dpi=110, bbox_inches="tight")
    plt.close(fig)
    return path


picture = draw_report()
if picture:
    display(Image(picture))


# ---------------- (3) history of one account ----------------
def show_history(acc, n=15):
    """Past transactions of one account. (No timestamps in simulated data,
    so order = order recorded.)"""
    if acc not in feat.index:
        print("Account not found:", acc)
        return
    h = df_all[(df_all["nameOrig"] == acc) | (df_all["nameDest"] == acc)].copy()
    sent = h["nameOrig"] == acc
    h["Direction"] = np.where(sent, "SENT", "RECEIVED")
    h["Counterparty"] = np.where(sent, h["nameDest"], h["nameOrig"])
    h["Counterparty Type"] = h["Counterparty"].map(feat["type"]).fillna("Unknown")
    h["Counterparty Risk %"] = (h["Counterparty"].map(feat["risk"]).fillna(0) * 100).round(1)
    h["Amount"] = h["amount"].round(2)
    h["Fraud-linked?"] = np.where(h["isFraud"] == 1, "YES", "no")
    cols = ["Direction", "Counterparty", "Counterparty Type",
            "Counterparty Risk %", "Amount", "Fraud-linked?"]
    r = feat.loc[acc]
    print(f"\nHISTORY OF {acc}  |  type: {r['type']}  |  risk: {r['risk'] * 100:.1f}%  |  "
          f"{int(r['tx_count'])} transactions in total (showing last {min(n, len(h))})")
    display(h[cols].tail(n).reset_index(drop=True))


if len(flagged_df):
    show_history(flagged_df.index[0])
    print("\nTo see any other account, run e.g.:  show_history('" + flagged_df.index[1] + "')"
          if len(flagged_df) > 1 else "")
```

```python
# ======================================================================
# CELL 4: MuleWatch website + live backend (run AFTER cells 1-3)
# First, in its own cell:   !pip install -q pyngrok
# Then paste your ngrok token below and run this cell.
# FIX in this version: the Check-account box can now be typed into
# (onkeydown handler no longer returns false).
# ======================================================================
import threading, time, random, re, os
import numpy as np, pandas as pd, uvicorn
from fastapi import FastAPI, HTTPException
from fastapi.responses import HTMLResponse, Response
from pydantic import BaseModel

NGROK_TOKEN = ""                                     # <- paste your ngrok authtoken
NGROK_DOMAIN = ""  # your static domain (or "" for a random one)
PORT = 8000

LOCK = threading.RLock()
MODEL = STATE["model"]
LEDGER = df_all[["nameOrig", "nameDest", "amount", "isFraud"]].copy()
LEDGER["ts"] = np.nan                       # history has no timestamps; live transfers do
LEDGER = LEDGER.reset_index(drop=True)
FEAT = feat.copy()                          # from cell 3 (has risk, type, fraud_txns...)
FROZEN, EVENTS, NEXT = set(), [], [1]
NAMES = ["Aarav", "Vivaan", "Diya", "Isha", "Karthik", "Meena", "Rohan", "Priya", "Arjun", "Sneha", "Farhan", "Lakshmi"]
SURN = ["Kumar", "Sharma", "Iyer", "Reddy", "Nair", "Khan", "Das", "Patel", "Singh", "Menon"]
CITY = ["Chennai", "Mumbai", "Delhi", "Bengaluru", "Hyderabad", "Kolkata", "Pune", "Kochi"]
BANK = ["SBI", "HDFC Bank", "ICICI Bank", "Axis Bank", "Canara Bank", "Indian Bank"]
BIZ = ["Sri Murugan Traders", "Annapoorna Foods", "Lotus Pharmacy", "Metro Electronics", "Bharat Logistics",
       "Green Valley School", "City Care Clinic", "Urban Threads", "QuickMart", "Sunrise Stationers"]


def is_mule_like(a):
    return FEAT.loc[a, "risk"] > 0.5


def reasons(a):
    r, out = FEAT.loc[a], []
    if r["total_in"] > 0 and r["pass_through"] >= 0.9:
        out.append(f"Passes on {r['pass_through'] * 100:.0f}% of the money it receives")
    L = LEDGER[(LEDGER.nameOrig == a) | (LEDGER.nameDest == a)]
    L = L[L.ts.notna()]
    ins, outs = L[L.nameDest == a], L[L.nameOrig == a]
    if len(ins) and len(outs) and (outs.ts.min() - ins.ts.min()) < 3600:
        out.append("Most outgoing money was forwarded within an hour of arriving")
    if r["in_degree"] <= 2 and r["out_degree"] <= 2 and r["total_out"] >= 5000:
        out.append("Large amounts go to very few accounts")
    if r["community_size"] <= 15:
        out.append(f"Sits in a tight cluster of {int(r['community_size'])} accounts")
    if r["fraud_txns"] > 0:
        out.append(f"{int(r['fraud_txns'])} fraud-linked transaction(s) in its history")
    if len(L) > 1 and (L.ts.max() - L.ts.min()) < 7200:
        out.append("Very active for a short time")
    if r["risk"] >= 0.5 and not out:
        out.append("Overall transfer pattern closely matches known mule accounts (model score above 50%)")
    return out if r["risk"] >= 0.3 else []


def profile(a):
    g = random.Random(a)
    biz = FEAT.loc[a, "type"] == "Business / Seller" if a in FEAT.index else False
    mule = a in FEAT.index and (FEAT.loc[a, "fraud_txns"] > 0 or a.startswith("LIVE_"))
    name = BIZ[sum(map(ord, a)) % len(BIZ)] if biz else f"{g.choice(NAMES)} {g.choice(SURN)}"
    days = g.randint(4, 60) if mule else g.randint(300, 3500)
    return {"name": name, "fields": {
        "Account type": "Business" if biz else "Personal", "City": g.choice(CITY), "Bank": g.choice(BANK),
        "KYC": g.choice(["Minimal KYC", "Pending re-KYC"]) if mule else "Full KYC verified",
        "Opened": f"{days} days ago", "Occupation": g.choice(["Student", "Delivery partner", "Daily wage worker"]) if mule else g.choice(["Engineer", "Teacher", "Shop owner", "Accountant", "Nurse"]),
        "Phone": f"+91 9{g.randint(10, 99)}•• ••{g.randint(100, 999)}", "PAN": f"{g.choice('ABCDEFG')}{g.choice('KLMNPQR')}•••••{g.randint(1, 9)}{g.choice('ABCDEFG')}"}}


def rescore(accs):
    """Recompute features for given accounts from the ledger and score them with the cell-1 model."""
    for a in accs:
        i, o = LEDGER[LEDGER.nameDest == a], LEDGER[LEDGER.nameOrig == a]
        new = a not in FEAT.index
        if new:
            FEAT.loc[a] = 0
            FEAT.loc[a, ["community_size", "type", "label"]] = [3, "Personal", 0]
        FEAT.loc[a, "total_in"], FEAT.loc[a, "total_out"] = i.amount.sum(), o.amount.sum()
        FEAT.loc[a, "in_degree"], FEAT.loc[a, "out_degree"] = i.nameOrig.nunique(), o.nameDest.nunique()
        FEAT.loc[a, "pass_through"] = float(make_pass_through(i.amount.sum(), o.amount.sum()))
        FEAT.loc[a, "tx_count"] = len(i) + len(o)
        x = FEAT.loc[[a], FEATURES].astype(float)
        FEAT.loc[a, "risk"] = float(MODEL.predict_proba(x)[0][1])


def event(a, kind, why=""):
    EVENTS.insert(0, {"id": NEXT[0], "ts": time.time(), "acct": a, "event": kind,
                      "score": float(FEAT.loc[a, "risk"]), "reason": why or (reasons(a) or ["Model score"])[0]})
    NEXT[0] += 1


def add_tx(src, dst, amt):
    if not (re.fullmatch(r"[A-Za-z0-9_\-]{1,24}", src or "") and re.fullmatch(r"[A-Za-z0-9_\-]{1,24}", dst or "")):
        raise HTTPException(400, "Account IDs may use letters, numbers, _ and - only")
    if src == dst:
        raise HTTPException(400, "Sender and receiver must differ")
    if not (0 < amt <= 1e8):
        raise HTTPException(400, "Enter a valid amount")
    if src in FROZEN or dst in FROZEN:
        raise HTTPException(409, "Transfer BLOCKED: an account involved is frozen")
    with LOCK:
        before = {a: float(FEAT.loc[a, "risk"]) if a in FEAT.index else 0.0 for a in (src, dst)}
        LEDGER.loc[len(LEDGER)] = [src, dst, float(amt), 0, time.time()]
        rescore([src, dst])
        for a in (src, dst):
            was, now = before[a] > 0.5, FEAT.loc[a, "risk"] > 0.5
            if now and not was:
                event(a, "FLAGGED")
            if was and not now:
                event(a, "CLEARED")
        return {"src": src, "dst": dst, "src_before": before[src], "dst_before": before[dst],
                "src_risk": float(FEAT.loc[src, "risk"]), "dst_risk": float(FEAT.loc[dst, "risk"]),
                "flagged": bool(FEAT.loc[dst, "risk"] > 0.5)}


# first flags (history) so the flag-history page is not empty
for _a in FEAT.index[FEAT.risk > 0.5][:60]:
    EVENTS.append({"id": 0, "ts": 0, "acct": _a, "event": "FLAGGED", "score": float(FEAT.loc[_a, "risk"]),
                   "reason": (reasons(_a) or ["Model score"])[0]})

SIM = {"chain": [], "step": 0, "amt": 0, "n": 0}
mw = FastAPI(title="MuleWatch")


class Tx(BaseModel):
    src: str
    dst: str
    amount: float


@mw.get("/", response_class=HTMLResponse)
def home():
    return HTML


@mw.get("/api/search")
def search(q: str = ""):
    q = q.lower()
    return [a for a in FEAT.index if q in a.lower()][:12]


@mw.get("/api/summary")
def summary():
    with LOCK:
        fl = FEAT[(FEAT.risk > 0.5) & ~FEAT.index.isin(FROZEN)]
        flagged_set = set(FEAT.index[FEAT.risk > 0.5])
        rec = LEDGER.tail(14).iloc[::-1]
        top = FEAT[FEAT.risk > 0.5].sort_values("risk", ascending=False).head(12)
        return {"transfers": len(LEDGER), "accounts": len(FEAT), "flagged": len(flagged_set), "frozen": len(FROZEN),
                "volume": float(LEDGER.amount.sum()),
                "flagged_volume": float(LEDGER[LEDGER.nameOrig.isin(flagged_set) | LEDGER.nameDest.isin(flagged_set)].amount.sum()),
                "hist": np.histogram(FEAT.risk, bins=10, range=(0, 1))[0].tolist(),
                "model": {"precision": MODEL_METRICS[0], "recall": MODEL_METRICS[1]},
                "recent": [{"id": int(i), "src": r.nameOrig, "dst": r.nameDest, "amt": float(r.amount),
                            "ts": None if pd.isna(r.ts) else float(r.ts), "to_risk": float(FEAT.loc[r.nameDest, "risk"]),
                            "why": (reasons(r.nameDest) or [""])[0] if FEAT.loc[r.nameDest, "risk"] > 0.5 else ""}
                           for i, r in rec.iterrows()],
                "alerts": EVENTS[:5],
                "top": [{"id": a, "name": profile(a)["name"], "risk": float(r.risk), "reasons": reasons(a) or ["Model score above 50%"]}
                        for a, r in top.iterrows()]}


@mw.get("/api/account/{a}")
def account(a: str):
    with LOCK:
        if a not in FEAT.index:
            raise HTTPException(404, "Account not found in the dataset")
        r = FEAT.loc[a]
        L = LEDGER[(LEDGER.nameOrig == a) | (LEDGER.nameDest == a)].tail(40).iloc[::-1]
        now = time.time()
        hist = []
        for _, t in L.iterrows():
            sent = t.nameOrig == a
            o = t.nameDest if sent else t.nameOrig
            hist.append({"dir": "SENT" if sent else "GOT", "other": o, "amt": float(t.amount), "fraud": bool(t.isFraud),
                         "ts": None if pd.isna(t.ts) else float(t.ts), "other_risk": float(FEAT.loc[o, "risk"]) if o in FEAT.index else 0})
        nb = {h["other"] for h in hist}
        links = sorted([(o, float(FEAT.loc[o, "risk"])) for o in nb if o in FEAT.index and FEAT.loc[o, "risk"] > 0.5], key=lambda x: -x[1])
        lh = LEDGER[(LEDGER.ts.notna()) & (now - LEDGER.ts < 3600)]
        return {"id": a, "risk": float(r.risk), "type": r["type"], "frozen": a in FROZEN, "profile": profile(a),
                "reasons": reasons(a), "signals": [[n, float(i), f"{float(r[n]):,.2f}"] for n, i in sorted(zip(FEATURES, MODEL.feature_importances_), key=lambda t: -t[1])[:4]], "history": hist, "links": links[:8],
                "events": [e for e in EVENTS if e["acct"] == a][:6],
                "numbers": [["Transfers", int(r.tx_count)], ["Total received", f"₹{r.total_in:,.0f}"], ["Total sent", f"₹{r.total_out:,.0f}"],
                            ["Share of received money sent on", f"{r.pass_through * 100:.0f}%"],
                            ["Sent in the last hour", f"₹{lh[lh.nameOrig == a].amount.sum():,.0f}"],
                            ["Received in the last hour", f"₹{lh[lh.nameDest == a].amount.sum():,.0f}"],
                            ["Different accounts involved", int(r.in_degree + r.out_degree)],
                            ["Cluster size", int(r.community_size)]]}


@mw.post("/api/transfer")
def transfer(t: Tx):
    return add_tx(t.src.strip(), t.dst.strip(), t.amount)


@mw.post("/api/freeze/{a}")
def freeze(a: str):
    if a not in FEAT.index:
        raise HTTPException(404, "Account not found")
    with LOCK:
        FROZEN.symmetric_difference_update({a})
        event(a, "FROZEN" if a in FROZEN else "UNFROZEN", "Action taken by analyst")
    return {"frozen": a in FROZEN}


@mw.get("/api/orgs")
def orgs(q: str = ""):
    B = FEAT[FEAT.type == "Business / Seller"]
    out = []
    for a, r in B.iterrows():
        nm = profile(a)["name"]
        if q.lower() in (nm + a).lower():
            nb = set(LEDGER[LEDGER.nameDest == a].nameOrig) | set(LEDGER[LEDGER.nameOrig == a].nameDest)
            out.append({"id": a, "name": nm, "risk": float(r.risk), "total_in": float(r.total_in), "total_out": float(r.total_out),
                        "in_degree": int(r.in_degree), "flagged_links": int(sum(1 for o in nb if o in FEAT.index and FEAT.loc[o, "risk"] > 0.5))})
    return out[:30]


@mw.get("/api/datasets")
def datasets():
    base = LEDGER[LEDGER.ts.isna()]
    paysim = os.path.exists(PAYSIM_FILE)
    return {"sources": [
        {"name": "PaySim (Kaggle, simulated mobile money)" if paysim else "Generated demo dataset",
         "role": "Used to train the model and shown as account history",
         "rows": len(base), "accounts": int(pd.concat([base.nameOrig, base.nameDest]).nunique()),
         "fraud_rows": int(base.isFraud.sum()), "columns": "nameOrig, nameDest, amount, isFraud"},
        {"name": "Live transfers", "role": "Recorded in this website while it runs",
         "rows": len(LEDGER) - len(base), "accounts": None, "fraud_rows": None, "columns": "same + timestamp"},
        {"name": "Elliptic (real Bitcoin, labelled)", "role": "Separate accuracy-test cell. Not loaded into this site",
         "rows": None, "accounts": None, "fraud_rows": None, "columns": "166 features per transaction"}]}


@mw.get("/api/dataset/rows")
def dataset_rows(q: str = "", fraud: int = 0, offset: int = 0, limit: int = 20):
    with LOCK:
        L = LEDGER
        if q:
            L = L[L.nameOrig.str.contains(q, case=False, regex=False) | L.nameDest.str.contains(q, case=False, regex=False)]
        if fraud:
            L = L[L.isFraud == 1]
        page = L.iloc[max(0, offset):max(0, offset) + min(limit, 100)]
        return {"total": len(L), "rows": [
            {"i": int(i), "src": r.nameOrig, "dst": r.nameDest, "amt": float(r.amount), "fraud": bool(r.isFraud),
             "live": not pd.isna(r.ts), "src_risk": float(FEAT.loc[r.nameOrig, "risk"]),
             "dst_risk": float(FEAT.loc[r.nameDest, "risk"])} for i, r in page.iterrows()]}


@mw.get("/api/dataset/download")
def dataset_download():
    with LOCK:
        csv = LEDGER.to_csv(index=False)
    return Response(csv, media_type="text/csv", headers={"Content-Disposition": "attachment; filename=mulewatch_transactions.csv"})


@mw.get("/api/events")
def events():
    return EVENTS[:80]


@mw.post("/api/simulate")
def simulate():
    """Creates live traffic: mostly normal payments, sometimes a fast mule chain (LIVE_*)."""
    with LOCK:
        normal = [a for a in FEAT.index if a.startswith("C")]
        if random.random() < 0.35:
            if not SIM["chain"]:
                SIM.update(chain=[f"LIVE_{SIM['n']}_{i}" for i in range(4)], step=0, amt=random.randint(30000, 90000), n=SIM["n"] + 1)
            c, s = SIM["chain"], SIM["step"]
            if s == 0:
                add_tx(random.choice(normal), c[0], SIM["amt"])
            elif s < len(c):
                SIM["amt"] *= 0.97
                add_tx(c[s - 1], c[s], round(SIM["amt"], 2))
            else:
                add_tx(c[-1], random.choice(normal), round(SIM["amt"] * 0.97, 2))
                SIM["chain"] = []
            SIM["step"] = s + 1 if SIM["chain"] else 0
        else:
            a, b = random.sample(normal, 2)
            add_tx(a, b, round(random.lognormvariate(6.5, 1.0), 2))
    return {"ok": True}

HTML = r"""<!DOCTYPE html>
<html lang="en"><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1">
<title>MuleWatch | Mule Account Detection</title>
<link href="https://fonts.googleapis.com/css2?family=Orbitron:wght@600;800&family=JetBrains+Mono:wght@400;600&family=Inter:wght@400;600&display=swap" rel="stylesheet">
<style>
:root{--bg:#05060a;--p:rgba(20,10,14,.72);--r:#ff2a4d;--r2:#ff6b81;--t:#ecedf2;--m:#8a8c99;--ok:#4be3a0;--b:rgba(255,42,77,.28)}
*{box-sizing:border-box}html,body{margin:0;background:var(--bg);color:var(--t);font:14px Inter,system-ui,sans-serif}
body:before{content:"";position:fixed;inset:0;z-index:-2;background:radial-gradient(900px 500px at 80% -10%,rgba(255,42,77,.22),transparent),linear-gradient(rgba(255,42,77,.06) 1px,transparent 1px) 0 0/44px 44px,linear-gradient(90deg,rgba(255,42,77,.06) 1px,transparent 1px) 0 0/44px 44px;animation:grid 14s linear infinite}
body:after{content:"";position:fixed;left:0;right:0;height:120px;top:-120px;z-index:-1;background:linear-gradient(transparent,rgba(255,42,77,.07),transparent);animation:scan 6s linear infinite}
@keyframes grid{to{background-position:0 0,0 44px,44px 0}}@keyframes scan{to{top:100%}}
@keyframes pulse{0%,100%{box-shadow:0 0 0 0 rgba(255,42,77,.7)}50%{box-shadow:0 0 0 8px rgba(255,42,77,0)}}
@keyframes slide{from{opacity:0;transform:translateX(-18px)}to{opacity:1;transform:none}}
@keyframes glitch{0%,92%,100%{text-shadow:0 0 12px var(--r)}94%{text-shadow:3px 0 var(--r),-3px 0 #fff}96%{text-shadow:-3px 0 var(--r),3px 0 #fff}}
@keyframes rise{from{opacity:0;transform:translateY(14px)}to{opacity:1;transform:none}}
nav{display:flex;align-items:center;gap:6px;padding:12px 26px;border-bottom:1px solid var(--b);backdrop-filter:blur(10px);position:sticky;top:0;z-index:9;background:rgba(5,6,10,.8)}
.logo{font:800 20px Orbitron;letter-spacing:3px;margin-right:22px;animation:glitch 5s infinite}.logo b{color:var(--r)}
nav a{color:var(--m);text-decoration:none;padding:8px 14px;border-radius:8px;font-weight:600;transition:.2s}nav a:hover,nav a.on{color:#fff;background:rgba(255,42,77,.16);box-shadow:inset 0 -2px var(--r)}
.sp{flex:1}.sw{display:flex;gap:8px;align-items:center;color:var(--m);font-size:12px;cursor:pointer;user-select:none}
.dot{width:9px;height:9px;border-radius:50%;background:var(--r);animation:pulse 1.6s infinite}.off .dot{background:#555;animation:none}
main{max-width:1280px;margin:0 auto;padding:26px;animation:rise .5s}
h1{font:800 22px Orbitron;margin:0 0 4px;letter-spacing:1px}.sub{color:var(--m);margin:0 0 20px}
.card{background:var(--p);border:1px solid var(--b);border-radius:14px;padding:18px;backdrop-filter:blur(8px);box-shadow:0 0 30px rgba(255,42,77,.06);transition:.25s}
.card:hover{border-color:var(--r);box-shadow:0 0 28px rgba(255,42,77,.2)}
.grid{display:grid;gap:16px}.k6{grid-template-columns:repeat(auto-fit,minmax(170px,1fr))}.two{grid-template-columns:2fr 1fr}.eq{grid-template-columns:1fr 1fr}
@media(max-width:900px){.two,.eq{grid-template-columns:1fr}}
.kpi small{color:var(--m);text-transform:uppercase;letter-spacing:1px;font-size:11px}.kpi div{font:600 26px 'JetBrains Mono',monospace;margin-top:6px}.red{color:var(--r);text-shadow:0 0 14px rgba(255,42,77,.6)}
h3{margin:0 0 12px;font:600 12px Orbitron;letter-spacing:2px;color:var(--r2);text-transform:uppercase}
table{width:100%;border-collapse:collapse;font-size:13px}th{color:var(--m);text-align:left;font-size:11px;letter-spacing:1px;text-transform:uppercase;padding:8px}td{padding:9px 8px;border-top:1px solid rgba(255,255,255,.06)}
tr.new{animation:slide .5s}tr.fl{background:rgba(255,42,77,.09)}tbody tr:hover{background:rgba(255,42,77,.14)}
.mono{font-family:'JetBrains Mono',monospace}.chip{padding:3px 9px;border-radius:20px;font-size:11px;font-weight:600;border:1px solid}
.HIGH,.MEDIUM,.LOW{color:var(--r);border-color:var(--r);background:rgba(255,42,77,.12)}.HIGH{box-shadow:0 0 10px rgba(255,42,77,.6)}.MEDIUM{opacity:.85}.LOW{opacity:.7}.CLEAR{color:var(--ok);border-color:rgba(75,227,160,.4)}.FROZEN{color:#fff;background:var(--r);border-color:var(--r)}
.bar{height:7px;border-radius:5px;background:rgba(255,255,255,.08);overflow:hidden;min-width:90px}.bar i{display:block;height:100%;background:linear-gradient(90deg,#7a1020,var(--r));width:0;transition:width 1s cubic-bezier(.2,.8,.2,1)}
input,select{background:#0b0c12;border:1px solid var(--b);color:#fff;padding:11px 13px;border-radius:9px;font:inherit;outline:none;width:100%;transition:.2s}input:focus{border-color:var(--r);box-shadow:0 0 14px rgba(255,42,77,.35)}
button{background:linear-gradient(135deg,#ff2a4d,#b3122d);color:#fff;border:0;padding:11px 20px;border-radius:9px;font:600 13px Inter;cursor:pointer;letter-spacing:.5px;transition:.2s}button:hover{transform:translateY(-1px);box-shadow:0 0 22px rgba(255,42,77,.55)}button.g{background:transparent;border:1px solid var(--b);color:var(--t)}
.row{display:flex;gap:12px;flex-wrap:wrap;align-items:end}.row>*{flex:1;min-width:150px}.row>button{flex:0 0 auto}
.nums{display:grid;grid-template-columns:1fr 1fr;gap:10px}.nums div{border:1px solid rgba(255,255,255,.07);border-radius:10px;padding:10px}.nums small{display:block;color:var(--m);font-size:11px}.nums b{font:600 16px 'JetBrains Mono'}
.gauge{position:relative;width:170px;height:170px;margin:auto}.gauge svg{transform:rotate(-90deg)}.gauge span{position:absolute;inset:0;display:grid;place-items:center;font:800 32px Orbitron}
.why li{margin:6px 0;color:#ffd0d8}.why li:before{content:"▸ ";color:var(--r)}
.ban{padding:12px 16px;border-radius:10px;margin-bottom:14px;border:1px solid var(--r);background:rgba(255,42,77,.14);animation:pulse 1.5s 3}
.toast{position:fixed;right:22px;bottom:22px;z-index:20;display:grid;gap:10px}.toast div{background:#12060a;border:1px solid var(--r);border-left:4px solid var(--r);padding:12px 16px;border-radius:10px;animation:slide .4s;box-shadow:0 0 24px rgba(255,42,77,.4);max-width:340px}
.mut{color:var(--m)}a.l{color:var(--r2);cursor:pointer;text-decoration:none}a.l:hover{text-decoration:underline}
</style></head><body>
<nav><div class="logo">MULE<b>WATCH</b></div><a href="#dash" id="n-dash">Dashboard</a><a href="#check" id="n-check">Check account</a><a href="#xfer" id="n-xfer">Record transfer</a><a href="#orgs" id="n-orgs">Organisations</a><a href="#flags" id="n-flags">Flag history</a><a href="#data" id="n-data">Datasets</a><div class="sp"></div><div class="sw" id="sim"><span class="dot"></span><span id="simt">LIVE SIMULATION ON</span></div></nav>
<main id="v"></main><div class="toast" id="toast"></div>
<script>
const $=s=>document.querySelector(s),esc=s=>String(s??'').replace(/[&<>"']/g,c=>({'&':'&amp;','<':'&lt;','>':'&gt;','"':'&quot;',"'":'&#39;'}[c]));
const inr=n=>'₹'+Number(n||0).toLocaleString('en-IN',{maximumFractionDigits:0}),pct=n=>(n*100).toFixed(0)+'%';
const api=async(p,o)=>{const r=await fetch(p,o);const j=await r.json().catch(()=>({}));if(!r.ok)throw new Error(j.detail||r.status);return j};
const lv=r=>r>=.85?'HIGH':r>=.65?'MEDIUM':r>=.5?'LOW':'CLEAR';
const chip=(r,fr)=>fr?'<span class="chip FROZEN">FROZEN</span>':`<span class="chip ${lv(r)}">${lv(r)==='CLEAR'?'CLEAR':'🚩 '+lv(r)} ${pct(r)}</span>`;
const tm=t=>t?new Date(t*1000).toLocaleTimeString('en-IN'):'—';
let T,SIM=localStorage.sim!=='0',seen=new Set(),last=0;
function toast(m){const d=document.createElement('div');d.innerHTML=m;$('#toast').append(d);setTimeout(()=>d.remove(),5000)}
function count(el,to,f=x=>Math.round(x).toLocaleString('en-IN')){const s=performance.now(),a=+el.dataset.v||0;el.dataset.v=to;(function st(n){const p=Math.min(1,(n-s)/700);el.textContent=f(a+(to-a)*p);if(p<1)requestAnimationFrame(st)})(s)}
function gauge(r){const c=2*Math.PI*70;return `<div class="gauge"><svg width="170" height="170"><circle cx="85" cy="85" r="70" fill="none" stroke="rgba(255,255,255,.08)" stroke-width="12"/><circle id="gc" cx="85" cy="85" r="70" fill="none" stroke="#ff2a4d" stroke-width="12" stroke-linecap="round" stroke-dasharray="${c}" stroke-dashoffset="${c}" style="transition:1.2s;filter:drop-shadow(0 0 8px #ff2a4d)"/></svg><span class="${r>=.5?'red':''}">${pct(r)}</span></div>`}
function anim(r){const c=2*Math.PI*70,g=$('#gc');if(g)setTimeout(()=>g.style.strokeDashoffset=c*(1-r),50)}
function setSim(){$('#sim').className='sw'+(SIM?'':' off');$('#simt').textContent=SIM?'LIVE SIMULATION ON':'LIVE SIMULATION OFF';localStorage.sim=SIM?1:0}
$('#sim').onclick=()=>{SIM=!SIM;setSim()};setSim();
setInterval(()=>{if(SIM)api('/api/simulate',{method:'POST'}).catch(()=>{})},3000);
function nav(){clearInterval(T);const h=location.hash.slice(1)||'dash',[p,a]=h.split('/');document.querySelectorAll('nav a').forEach(x=>x.classList.toggle('on',x.id==='n-'+p));({dash,check,xfer,orgs,flags,data:datasets})[p]?.(decodeURIComponent(a||''))}
addEventListener('hashchange',nav);
/* ---------- Dashboard ---------- */
function dash(){$('#v').innerHTML=`<h1>Command Dashboard</h1><p class="sub">Live view of transfers and accounts flagged as possible money mules.</p>
<div class="grid k6" id="kp"></div><br><div class="grid two"><div class="card"><h3>Live transaction stream</h3><table><thead><tr><th>Time</th><th>From</th><th>To</th><th>Amount</th><th>Receiver risk</th></tr></thead><tbody id="st"></tbody></table></div>
<div class="card"><h3>Risk distribution</h3><div id="hs"></div><br><h3>Model (simulated data)</h3><div class="mono mut" id="md"></div></div></div><br>
<div class="card"><h3>Flagged accounts</h3><table><thead><tr><th>Account</th><th>Holder</th><th>Risk</th><th>Why flagged</th><th></th></tr></thead><tbody id="fl"></tbody></table></div>`;
const tick=async()=>{const s=await api('/api/summary');
const K=[['Transfers',s.transfers],['Accounts',s.accounts],['Flagged',s.flagged,1],['Volume',s.volume,0,1],['Flagged volume',s.flagged_volume,1,1],['Frozen',s.frozen]];
if(!$('#kp').children.length)$('#kp').innerHTML=K.map(k=>`<div class="card kpi"><small>${k[0]}</small><div class="${k[2]?'red':''}"><span></span></div></div>`).join('');
K.forEach((k,i)=>count($('#kp').children[i].querySelector('span'),k[1],k[3]?inr:undefined));
$('#hs').innerHTML=s.hist.map((c,i)=>`<div style="display:flex;gap:8px;align-items:center;margin:4px 0"><small class="mono mut" style="width:54px">${i*10}-${i*10+10}%</small><div class="bar" style="flex:1"><i style="width:${Math.min(100,c/Math.max(...s.hist)*100)}%"></i></div><small class="mono">${c}</small></div>`).join('');
$('#md').innerHTML=`precision ${pct(s.model.precision)} · recall ${pct(s.model.recall)}<br>Trained on generated data.<br>Not a real-world accuracy.`;
$('#st').innerHTML=s.recent.map(t=>{const n=!seen.has(t.id);seen.add(t.id);return `<tr class="${n&&last?'new':''} ${t.to_risk>=.5?'fl':''}"><td class="mono">${tm(t.ts)}</td><td><a class="l" href="#check/${esc(t.src)}">${esc(t.src)}</a></td><td><a class="l" href="#check/${esc(t.dst)}">${esc(t.dst)}</a></td><td class="mono">${inr(t.amt)}</td><td>${chip(t.to_risk)}${t.why?`<div class="mut" style="font-size:11px;max-width:230px">${esc(t.why)}</div>`:""}</td></tr>`}).join('');
s.alerts.filter(a=>a.id>last&&last).forEach(a=>toast(`🚩 <b>${esc(a.acct)}</b> ${a.event}<br><small>${esc(a.reason)}</small>`));last=s.alerts[0]?s.alerts[0].id:last;
$('#fl').innerHTML=s.top.map(a=>`<tr class="fl"><td class="mono">${esc(a.id)}</td><td>${esc(a.name)}</td><td><div style="display:flex;gap:8px;align-items:center"><div class="bar"><i style="width:${a.risk*100}%"></i></div>${pct(a.risk)}</div></td><td class="mut">${a.reasons.map(r=>"▸ "+esc(r)).join("<br>")}</td><td><a class="l" href="#check/${esc(a.id)}">Details →</a></td></tr>`).join('')};
tick();T=setInterval(()=>tick().catch(()=>{}),2000)}
/* ---------- Check account ---------- */
function check(id){$('#v').innerHTML=`<h1>Check an Account</h1><p class="sub">Enter an account ID to see its risk score, holder details, reasons and transfer history.</p><div class="card"><div class="row"><div><small class="mut">Account ID</small><input id="q" list="dl" placeholder="e.g. M4_0, LIVE_B, C12, B3"></div><button id="go">Check risk</button></div><datalist id="dl"></datalist></div><br><div id="res"></div>`;
$('#q').oninput=async e=>{if(e.target.value.length>0)$('#dl').innerHTML=(await api('/api/search?q='+encodeURIComponent(e.target.value))).map(x=>`<option value="${esc(x)}">`).join('')};
const run=async()=>{const v=$('#q').value.trim();if(!v)return;history.replaceState(0,'','#check/'+encodeURIComponent(v));try{const a=await api('/api/account/'+encodeURIComponent(v));
$('#res').innerHTML=`${a.risk>=.5?'<div class="ban">🚩 <b>RED FLAG</b> — this account is flagged as a possible money mule and should be reviewed.</div>':''}
<div class="grid eq"><div class="card"><h3>Account holder</h3><div style="font:800 20px Orbitron">${esc(a.profile.name)}</div><div class="mono mut">${esc(a.id)}</div><p>${chip(a.risk,a.frozen)} <span class="chip CLEAR" style="color:var(--m);border-color:#333">${esc(a.type)}</span></p>
<div class="nums">${Object.entries(a.profile.fields).map(([k,v])=>`<div><small>${esc(k)}</small><b style="font-size:13px">${esc(v)}</b></div>`).join('')}</div><br><button class="g" id="fz">${a.frozen?'Unfreeze account':'Freeze account'}</button></div>
<div class="card"><h3>Risk score</h3>${gauge(a.risk)}<ul class="why">${a.reasons.map(r=>`<li>${esc(r)}</li>`).join('')||'<li class="mut">No suspicious pattern detected</li>'}</ul></div></div><br>
<div class="card"><h3>Key numbers <span class="mut" id="lv"></span></h3><div class="nums" style="grid-template-columns:repeat(auto-fit,minmax(160px,1fr))">${a.numbers.map(n=>`<div><small>${esc(n[0])}</small><b>${esc(n[1])}</b></div>`).join('')}</div></div><br>
<div class="card" style="border-color:${a.risk>=.5?'var(--r)':'var(--b)'}"><h3>${a.risk>=.5?'Why was this account flagged?':'Why is this account not flagged?'}</h3><ol class="why" style="padding-left:20px">${a.reasons.map(r=>`<li>${esc(r)}</li>`).join('')||'<li class="mut">No suspicious pattern found in its transfers.</li>'}</ol><h3 style="margin-top:16px">Strongest signals the model uses (with this account's value)</h3>${a.signals.map(s=>`<div style="display:flex;gap:10px;align-items:center;margin:7px 0"><span class="mono mut" style="width:130px">${esc(s[0])}</span><div class="bar" style="flex:1"><i style="width:${s[1]/a.signals[0][1]*100}%"></i></div><b class="mono" style="width:110px;text-align:right">${esc(s[2])}</b></div>`).join('')}</div><br>
<div class="grid eq"><div class="card"><h3>Transfer history</h3><table><thead><tr><th>Time</th><th></th><th>Account</th><th>Amount</th><th>Risk</th></tr></thead><tbody>${a.history.map(h=>`<tr class="${h.fraud?'fl':''}"><td class="mono">${h.ts?tm(h.ts):'recorded'}</td><td>${h.dir==='SENT'?'→ sent':'← got'}</td><td><a class="l" href="#check/${esc(h.other)}">${esc(h.other)}</a></td><td class="mono">${inr(h.amt)}</td><td>${chip(h.other_risk)}</td></tr>`).join('')}</tbody></table></div>
<div class="card"><h3>Previous flag history</h3>${a.events.map(e=>`<p><span class="chip HIGH">${esc(e.event)}</span> <span class="mono">${pct(e.score)}</span> <span class="mut">${esc(e.reason)}</span></p>`).join('')||'<p class="mut">No previous flags.</p>'}<h3>Flagged accounts linked to it</h3>${a.links.map(l=>`<p><a class="l" href="#check/${esc(l[0])}">${esc(l[0])}</a> ${chip(l[1])}</p>`).join('')||'<p class="mut">None.</p>'}</div></div>`;
anim(a.risk);$('#fz').onclick=async()=>{await api('/api/freeze/'+encodeURIComponent(a.id),{method:'POST'});run()};
clearInterval(T);T=setInterval(()=>{const g=$('#gc');run2(a.id)},3000)}catch(e){$('#res').innerHTML=`<div class="card red">${esc(e.message)}</div>`}};
const run2=async i=>{if(location.hash.indexOf('check')<0)return;try{const a=await api('/api/account/'+encodeURIComponent(i));$('#lv').textContent='· live '+new Date().toLocaleTimeString('en-IN');}catch(e){}};
$('#go').onclick=run;$('#q').onkeydown=e=>{if(e.key==='Enter')run()};if(id){$('#q').value=id;run()}}
/* ---------- Record transfer ---------- */
function xfer(){$('#v').innerHTML=`<h1>Record a Transfer</h1><p class="sub">Add a transfer as it happens. Both accounts are re-scored instantly.</p><div class="grid eq"><div class="card"><div class="row"><div><small class="mut">From account</small><input id="f" list="dl" placeholder="e.g. LIVE_A"></div><div><small class="mut">To account</small><input id="t" list="dl" placeholder="e.g. LIVE_B"></div></div><br><div class="row"><div><small class="mut">Amount (₹)</small><input id="a" type="number" placeholder="e.g. 40000"></div><button id="s">Record transfer</button></div><datalist id="dl"></datalist><p class="mut">Tip: send ₹40,000 LIVE_A → LIVE_B, then LIVE_B → LIVE_C for most of it, and watch the risk climb.</p></div><div class="card" id="o"><h3>Result</h3><p class="mut">Submit a transfer to see the new risk scores.</p></div></div><br><div class="card"><h3>Recent transfers</h3><table><tbody id="rc"></tbody></table></div>`;
$('#f').oninput=$('#t').oninput=async e=>{$('#dl').innerHTML=(await api('/api/search?q='+encodeURIComponent(e.target.value))).map(x=>`<option value="${esc(x)}">`).join('')};
const rc=async()=>{$('#rc').innerHTML=(await api('/api/summary')).recent.slice(0,8).map(t=>`<tr><td class="mono">${tm(t.ts)}</td><td>${esc(t.src)} → ${esc(t.dst)}</td><td class="mono">${inr(t.amt)}</td><td>${chip(t.to_risk)}</td></tr>`).join('')};rc();T=setInterval(rc,3000);
$('#s').onclick=async()=>{try{const r=await api('/api/transfer',{method:'POST',headers:{'Content-Type':'application/json'},body:JSON.stringify({src:$('#f').value.trim(),dst:$('#t').value.trim(),amount:+$('#a').value})});
$('#o').innerHTML=`<h3>Result</h3>${r.flagged?'<div class="ban">🚩 <b>ALERT</b> receiver is now flagged</div>':''}${gauge(r.dst_risk)}<p style="text-align:center">${esc(r.dst)} <span class="mut">was ${pct(r.dst_before)}</span></p><p class="mut" style="text-align:center">Sender ${esc(r.src)}: ${pct(r.src_before)} → ${pct(r.src_risk)}</p>`;anim(r.dst_risk);rc()}catch(e){$('#o').innerHTML=`<h3>Result</h3><div class="ban">⛔ ${esc(e.message)}</div>`}}}
/* ---------- Organisations ---------- */
function orgs(){$('#v').innerHTML=`<h1>Organisation Records</h1><p class="sub">Business accounts, their money flow and flagged accounts they deal with.</p><div class="card"><input id="q" placeholder="Search organisation name or account ID..."></div><br><div class="grid" id="g" style="grid-template-columns:repeat(auto-fill,minmax(280px,1fr))"></div>`;
const ld=async()=>{const o=await api('/api/orgs?q='+encodeURIComponent($('#q').value));$('#g').innerHTML=o.map(x=>`<div class="card" style="cursor:pointer" onclick="location.hash='#check/${esc(x.id)}'"><div style="font:600 15px Orbitron">${esc(x.name)}</div><div class="mono mut">${esc(x.id)}</div><p>${chip(x.risk)}</p><div class="nums"><div><small>Received</small><b>${inr(x.total_in)}</b></div><div><small>Sent</small><b>${inr(x.total_out)}</b></div><div><small>Customers</small><b>${x.in_degree}</b></div><div><small>Flagged links</small><b class="${x.flagged_links?'red':''}">${x.flagged_links}</b></div></div></div>`).join('')};
$('#q').oninput=ld;ld()}
/* ---------- Flag history ---------- */
function flags(){$('#v').innerHTML=`<h1>Flag History</h1><p class="sub">Every time an account was flagged, cleared or frozen.</p><div class="card"><table><thead><tr><th>Time</th><th>Account</th><th>Event</th><th>Score</th><th>Reason</th></tr></thead><tbody id="e"></tbody></table></div>`;
const ld=async()=>{$('#e').innerHTML=(await api('/api/events')).map(e=>`<tr class="${e.event==='FLAGGED'?'fl':''}"><td class="mono">${e.ts?tm(e.ts):'earlier'}</td><td><a class="l" href="#check/${esc(e.acct)}">${esc(e.acct)}</a></td><td><span class="chip HIGH">${esc(e.event)}</span></td><td class="mono">${pct(e.score)}</td><td class="mut">${esc(e.reason)}</td></tr>`).join('')};ld();T=setInterval(ld,3000)}
function datasets(){$('#v').innerHTML=`<h1>Datasets</h1><p class="sub">The data behind every score. Browse the records the model learned from.</p><div class="grid" id="ds" style="grid-template-columns:repeat(auto-fit,minmax(280px,1fr))"></div><br><div class="card"><h3>Transaction records</h3><div class="row"><div><input id="q" placeholder="Filter by account ID, e.g. M4_0"></div><label class="sw"><input id="fo" type="checkbox" style="width:auto"> labelled fraud only</label><a href="/api/dataset/download"><button class="g">Download CSV</button></a></div><table><thead><tr><th>#</th><th>From</th><th>To</th><th>Amount</th><th>Label</th><th>Source</th><th>Sender risk</th><th>Receiver risk</th></tr></thead><tbody id="tb"></tbody></table><div class="row" style="margin-top:12px"><button class="g" id="pv">← Prev</button><span class="mut" id="pg"></span><button class="g" id="nx">Next →</button></div></div>`;
let off=0;const ld=async()=>{const r=await api(`/api/dataset/rows?q=${encodeURIComponent($('#q').value)}&fraud=${$('#fo').checked?1:0}&offset=${off}&limit=20`);
$('#tb').innerHTML=r.rows.map(x=>`<tr class="${x.fraud?'fl':''}"><td class="mono">${x.i}</td><td><a class="l" href="#check/${esc(x.src)}">${esc(x.src)}</a></td><td><a class="l" href="#check/${esc(x.dst)}">${esc(x.dst)}</a></td><td class="mono">${inr(x.amt)}</td><td>${x.fraud?'<span class="chip HIGH">FRAUD</span>':'normal'}</td><td class="mut">${x.live?'live':'dataset'}</td><td>${chip(x.src_risk)}</td><td>${chip(x.dst_risk)}</td></tr>`).join('');$('#pg').textContent=`${r.total?off+1:0}-${Math.min(off+20,r.total)} of ${r.total.toLocaleString('en-IN')}`;$('#pv').disabled=off==0;$('#nx').disabled=off+20>=r.total};
$('#q').oninput=$('#fo').onchange=()=>{off=0;ld()};$('#pv').onclick=()=>{off=Math.max(0,off-20);ld()};$('#nx').onclick=()=>{off+=20;ld()};ld();
api('/api/datasets').then(d=>{$('#ds').innerHTML=d.sources.map(s=>`<div class="card"><h3>${esc(s.name)}</h3><p class="mut">${esc(s.role)}</p><div class="nums"><div><small>Rows</small><b>${s.rows==null?'—':s.rows.toLocaleString('en-IN')}</b></div><div><small>Accounts</small><b>${s.accounts==null?'—':s.accounts.toLocaleString('en-IN')}</b></div><div><small>Fraud rows</small><b class="red">${s.fraud_rows==null?'—':s.fraud_rows.toLocaleString('en-IN')}</b></div><div><small>Columns</small><b style="font-size:11px">${esc(s.columns)}</b></div></div></div>`).join('')})}
nav();
</script></body></html>
"""


# ---------------- website + ngrok ----------------
from sklearn.metrics import precision_score, recall_score
_m = FEAT[FEATURES].astype(float)
MODEL_METRICS = (float(precision_score(FEAT.label, MODEL.predict(_m), zero_division=0)),
                 float(recall_score(FEAT.label, MODEL.predict(_m), zero_division=0)))
mw.mount("/legacy", app)   # your earlier API (/score, /flagged, /account) stays available at /legacy/docs

try:
    MW_SERVER.should_exit = True    # stop an older run of this cell
    time.sleep(1.5)
except NameError:
    pass
MW_SERVER = uvicorn.Server(uvicorn.Config(mw, host="0.0.0.0", port=PORT, log_level="warning"))
threading.Thread(target=MW_SERVER.run, daemon=True).start()
time.sleep(2)

if NGROK_TOKEN:
    from pyngrok import ngrok
    ngrok.kill()
    ngrok.set_auth_token(NGROK_TOKEN)
    tun = ngrok.connect(PORT, domain=NGROK_DOMAIN) if NGROK_DOMAIN else ngrok.connect(PORT)
    print("\nMULEWATCH IS LIVE ->", tun.public_url)
else:
    print("Server running inside Colab. Paste NGROK_TOKEN at the top and re-run to get a public link.")
```
